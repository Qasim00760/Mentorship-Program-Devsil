<h2 style="font-size:20px; font-family:Georgia, serif; color:#0D47A1; text-align: left; margin: 0;">
        1.1 Introduction
    </h2>
</div>


Employee attrition refers to the loss of employees from an organization due to resignation, retirement, or other reasons. Managing attrition is a critical task for Human Resource (HR) departments because high attrition can negatively affect organizational performance, increase operational costs, and disrupt workflow continuity.

With the advancement of Data Science and Machine Learning, organizations can now analyze employee data to identify patterns and predict which employees are likely to leave. Predictive analytics helps HR departments take proactive measures to improve employee satisfaction and retention.

In this project, we develop a Machine Learning-based Employee Attrition Prediction System using the IBM HR Analytics dataset. The system will analyze employee information such as salary, job satisfaction, work-life balance, and experience to predict attrition risk and provide valuable business insights.

<h2 style="font-size:20px; font-family:Georgia, serif; color:#0D47A1; text-align: left; margin: 0;">
        1.2 Business Problem
    </h2>
</div>


Employee attrition is a major challenge for organizations because it leads to several negative consequences, including:

- Increased recruitment and hiring costs

- Time and resources spent on training new employees

- Loss of experienced and skilled workers

- Reduced productivity and team performance

- Disruption in ongoing projects

The HR department currently lacks an automated system to identify employees who are at high risk of leaving the organization. Without predictive insights, HR can only react after employees resign rather than preventing attrition proactively.

Therefore, the organization needs a data-driven solution that can predict employee attrition in advance and identify the key factors contributing to attrition. This will help HR departments implement effective retention strategies and improve overall workforce stability.

```python
# Confidence interval for mean MonthlyIncome
import numpy as np
from scipy import stats

mean = df['MonthlyIncome'].mean()
std = df['MonthlyIncome'].std()
n = len(df)

confidence = 0.95
z = stats.norm.ppf((1 + confidence) / 2)

margin_error = z * (std / np.sqrt(n))

lower = mean - margin_error
upper = mean + margin_error

lower, upper
```

# Interpretation

| Actual \ Predicted | 0 (Stay) | 1 (Leave) |
| ------------------ | -------- | --------- |
| 0 (Stay)           | 52       | 203       |
| 1 (Leave)          | 4        | 35        |

Where:

- 0 = Employee stayed

- 1 = Employee left

Understanding the values

- True Negatives (TN) = 52

 - Employees who stayed and were correctly predicted to stay.

- False Positives (FP) = 203

 - Employees who stayed but the model predicted they would leave.

 - High number → Many “false alarms.”

- False Negatives (FN) = 4

 - Employees who left but the model predicted they would stay.

 - Very low → Model rarely misses actual leavers.

- True Positives (TP) = 35

 - Employees who left and were correctly predicted to leave.

# Interpretation

- The diagonal line represents a random classifier (AUC ≈ 0.5)

- Your curve is not far above the diagonal, which suggests:

 - The model has poor discriminative ability

 - It struggles to separate employees who will leave vs stay

- The fact that it reaches the top-right corner shows the model eventually predicts positives, but it does not do well at low false positive rates.

Key Metric

- AUC (Area Under the Curve) quantifies overall model performance:

 - 0.5 → Random

 - 0.7–0.8 → Fair

 - 0.8–0.9 → Good

 - 0.9+ → Excellent

Based on this curve, the AUC is likely around 0.55–0.6, indicating poor predictive power.

<h1>12. Final Model Selection</h1>
<p>HR Analytics | Machine Learning | Industry Project</p>
</div>

```python
# Model Comparison
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score

models = {
    "Logistic Regression": lr,
    "Decision Tree": dt,
    "Random Forest": rf,
    "SVM": svm,
    "KNN": knn
}

results = []

for name, model in models.items():
    y_pred = model.predict(X_test)
    
    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred)
    recall = recall_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)
    
    if hasattr(model, "predict_proba"):
        y_prob = model.predict_proba(X_test)[:,1]
        auc = roc_auc_score(y_test, y_prob)
    else:
        auc = None
        
    results.append([name, accuracy, precision, recall, f1, auc])

import pandas as pd
results_df = pd.DataFrame(results, columns=[
    "Model", "Accuracy", "Precision", "Recall", "F1 Score", "ROC-AUC"
])

results_df
```
